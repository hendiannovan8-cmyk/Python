**Global Variables**


```python
nama_aplikasi = "Aplikasi Kasir"  # Variabel GLOBAL


def cetak_header():
    print("=== " + nama_aplikasi + " ===")  # Membaca variabel global


def cetak_footer():
    print("Terima kasih telah menggunakan " + nama_aplikasi)


cetak_header()
print("Terima kasih telah menggunakan kami " + nama_aplikasi)
cetak_footer()
print("Terima kasih telah menggunakan saya " + nama_aplikasi)
# Baris bisa ditambahkan disini
```

**Local Variables**

```python

def sapa():
    pesan = "Halo dari dalam fungsi!"  # Variabel LOKAL
    print(pesan)


sapa()

# Baris di bawah ini akan menyebabkan Error (NameError)
```

**The Global Keywoard**


```python
# hanya mengubah nilai variabel dengan variable yang sama
skor = 0  # Variabel Global


def tambah_skor():
    global skor  # Menggunakan kata kunci 'global'
    skor = skor + 10  # Mengubah nilai variabel global


print("Skor awal:", skor)  # Output: 0
tambah_skor()
print("Skor setelah main:", skor)  # Output: 10
```

**Python Data Types**

```python
#Implicit (Penetapan Otomatis)

x = 5
d = 'halo'
f = 3.25
y = 1j

print(type(x))
print(type(d))
print(type(f))
print(type(y))
```
